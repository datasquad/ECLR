## Law of Large Numbers (LLN)

The Law of Large Numbers states that the sample mean converges to the population mean as the sample size increases, provided the population mean exists (is finite). We'll demonstrate this by sampling from a uniform and a normal distribution and observing how the sample means behave as we increase the sample size. 

First, we must define the distributions from which we will draw our random samples. We'll then repeatedly sample from these distributions and compute the sample means to observe how they converge toward the population mean $μ_x$. The Law of Large Numbers (LLN) clarifies that for a random variable with a finite mean $μ_x$, the average of $n$ random draws (i.e., the sample mean $\overline{X} = \frac{1}{n} \sum_{i=1}^{n} x_i$) will approximate $μ_x$ as $n$ increases.

This simulation involves setting up the distributions, drawing samples, and calculating their means to demonstrate the LLN in action visually.


```python
# Sample sizes
n = 10
# Parameters for the uniform distribution
lb, ub = 1, 2
# Parameters for the normal distribution
mu, std = 0, 1

# Generating random samples
x1 = np.random.uniform(lb, ub, n)
x2 = np.random.normal(mu, std, n)

# Calculating means
mean1 = np.mean(x1)
mean2 = np.mean(x2)

print("Sample 1 mean = ", mean1)
print("Sample 2 mean = ", mean2)
```

This code block initializes two distributions and calculates their sample means. The output reflects the calculated means, which will differ from the population means (1.5 for uniform and 0 for normal) due to the small sample size. The Law of Large Numbers (LLN) suggests that as the sample size increases indefinitely $( n \rightarrow \infty)$, the sample means should converge more closely to the population mean. Therefore, we need to code a loop that progressively increases the sample size $n$.

To see the LLN in action, we will increase the sample sizes.

```python
# Range of increasing sample sizes
n_sel = np.concatenate([np.arange(5, 51, 5), np.arange(60, 101, 10), np.arange(150, 5001, 50)])

# Containers for saving sample means
save_mean1 = []
save_mean2 = []

# Loop to calculate means for increasing sample sizes
for n in n_sel:
    save_mean1.append(np.mean(np.random.uniform(lb, ub, n)))
    save_mean2.append(np.mean(np.random.normal(mu, std, n)))

# Creating a DataFrame to hold results
results = pd.DataFrame({
    'n': n_sel,
    'mean1': save_mean1,
    'mean2': save_mean2
})
```

Running this code we have created a range of sample sizes and have calculated the means for each size, storing them for later visualization. Let's now plot the results using `lets-plot` to visualize the convergence of sample means to the population mean.

```python
# Add a column for the log of sample sizes
results['log_n'] = np.log(results['n'])

# Now plot using the new 'log_n' column
# Plotting sample means from the normal distribution
ggplot(results, aes('log_n', 'mean2')) + \
    geom_line() + \
    ggtitle('Sample means from normal distribution') + \
    xlab('Log of Sample Size') + \
    ylab('Sample Mean')
```

The titles, x-axis, and y-axis labels are set appropriately to convey this information. 

This visual representation is crucial for illustrating the Law of Large Numbers, showing that as the sample size increases, the sample means converge towards the population mean, a fundamental concept in statistics.

## Central Limit Theorem (CLT)


The Central Limit Theorem (CLT) states that under certain conditions, the sample mean of a sequence of random variables, $X_i$ (or $X_t$) when dealing with time series data), will approximate a normal distribution as the sample size becomes large, regardless of the original distribution of the variables. This remarkable property means that the sample mean is asymptotically normally distributed around the population mean $μ_x$, facilitating statistical inference across various contexts.

We will simulate a large population for $X$ to demonstrate the CLT in Python. Unlike our demonstration of the Law of Large Numbers, we will draw multiple samples of increasing sizes from this population. For each sample size, we'll calculate the sample mean for numerous subsets and then visualize how these sample means distribute themselves—highlighting the essence of the CLT.

Let's begin by creating a mixed population from two distributions. We'll generate 10,000 values from a uniform distribution (ranging between 1 and 2) and another 10,000 from a normal distribution. This setup will help us explore how the CLT operates across underlying data distributions. In this case, we can use libraries such as `numpy` for creating the random variables and handling array operations.


```python
import numpy as np

n = 10000
# set lower and upper bounds for uniform distribution
lb = 1
ub = 2
# set parameters for normal random variable
mu = 0
std = 1

# Generate random variables
x1 = np.random.uniform(low=lb, high=ub, size=n)  # Sampled from uniform distribution between 1 and 2
x2 = np.random.normal(loc=mu, scale=std, size=n)  # Sampled from normal distribution with mean 0 and sd = 1

# Combine both arrays into one population
Xpop = np.append(x1, x2)
```

We can now create a density plot using `lets-plot` library, to look at a distribution for this population.

This a clearly unusual population. Let’s also calculate the population **mean** and **standard deviation**.

To demonstrate the Central Limit Theorem using Python, we will simulate drawing 1000 samples each for two different sample sizes, $n_1=5$ and $n_2=30$, from the population we have. We will then calculate the sample means for each sample and plot the distributions of these sample means to observe how they behave as the sample size increases. 

```python
from lets_plot import *
LetsPlot.setup_html()

import numpy as np
import pandas as pd

# Sample sizes
n1 = 5
n2 = 30
num_samples = 1000

# Function to draw samples and compute their means
def sample_means(population, sample_size, num_samples):
    means = []
    for _ in range(num_samples):
        sample = np.random.choice(population, size=sample_size, replace=True)
        means.append(np.mean(sample))
    return means

# Calculate sample means for both sample sizes
means_n1 = sample_means(Xpop, n1, num_samples)
means_n2 = sample_means(Xpop, n2, num_samples)

# Convert to DataFrame for plotting
df_n1 = pd.DataFrame({'Sample Mean': means_n1, 'Sample Size': 'n=5'})
df_n2 = pd.DataFrame({'Sample Mean': means_n2, 'Sample Size': 'n=30'})
df = pd.concat([df_n1, df_n2])

# Plotting the distribution of sample means using lets-plot
p = ggplot(df, aes(x='Sample Mean', fill='Sample Size')) + \
    geom_histogram(alpha=0.7, bins=30, color='black') + \
    facet_grid(y='Sample Size') + \
    ggtitle("Distribution of Sample Means for Different Sample Sizes") + \
    xlab("Sample Mean") + ylab("Density") + \
    theme(legend_position='none')
p
```

From the two distribution plots we can see that:

- The larger the sample size ($𝑛$), the closer the distribution of sample means approximates a normal distribution, even if the underlying population distribution is not normal.
- For small sample sizes ($𝑛=5$), the sample means are more influenced by the shape of the population distribution.

To fully leverage the capabilities of `lets-plot`, it is best to organize the data in a long-format `DataFrame`.
```python
from lets_plot import *
LetsPlot.setup_html()

import numpy as np
import pandas as pd

# Sample sizes and repetitions
n_sel_clt = [5, 10, 30, 75]  # List of sample sizes
reps = 1000  # Number of repetitions

# Container to save results
save_mean5 = np.empty((reps, len(n_sel_clt)))  # Initialize an empty matrix

# Loop through sample sizes and repetitions
for j, n_clt in enumerate(n_sel_clt):
    for i in range(reps):
        save_mean5[i, j] = np.mean(np.random.choice(Xpop, size=n_clt, replace=True))

# Convert results to a Pandas DataFrame
col_names = [f"n{n:02d}" for n in n_sel_clt]
results_clt = pd.DataFrame(save_mean5, columns=col_names)

# Reshape the DataFrame to long format
results_clt_long = results_clt.melt(var_name='Sample.size', value_name='sm')

# Plotting the density plots for various sample sizes
p_clt = ggplot(results_clt_long, aes(x='sm', fill='Sample.size')) + \
    geom_density(alpha=0.2) + \
    ggtitle("Distribution of Sample Means for Various Sample Sizes") + \
    xlab("Sample Mean") + ylab("Density")
p_clt
```

You can easily indicate that as the sample size increases, the distributions of the sample means start to resemble a normal distribution, even though the population distribution is not normal. This is a direct manifestation of the CLT. Moreover, the variance of the sample mean distributions decreases as the sample size increases. This is expected because the variance of the sample mean,, $( \sigma_{\bar{x}}^2)$, is given by:

$$
\sigma_{\bar{x}}^2 = \frac{\sigma_X^2}{n}
$$

Here, $ \sigma_X^2$ is the population variance, and $n$ is the sample size. Thus, the variance decreases inversely with the sample size. The CLT states that the standardized sample mean  given by $Z = \frac{\bar{X} - \mu_X}{\sigma_{\bar{X}}}$, where $\sigma_{\bar{X}}$ is approximately standard normal $N(0, 1)$, and $\sigma_{\bar{X}} = \frac{\sigma_X}{\sqrt{n}}$


Moreover, when we know $μ_X$ (the population mean) and $sigma_X$ (the population standard deviation), we can standardize the sample mean directly. On the contrary, in real-world scenarios, $μ_X$) and $sigma_X$ are often unknown. Then:

1. We estimate $sigma_X$ with the sample standard deviation $s_X$.
2. The population mean $μ_X$ often comes from the null hypothesis for hypothesis testing.

### Standardised Sample Mean

When $μ_X$ (the population mean) and $sigma_X$ (the population standard deviation) are unknown, we work with:
$$
T = \frac{\bar{X} - \mu_X}{s_X / \sqrt{n}}
$$

This standardized sample mean is $t$-distributed with $n - 1$ degrees of freedom if $X$ is normally distributed.

If $X$ is not normally distributed, the Central Limit Theorem (CLT) still applies, and for sufficiently large $n$, $T$ will approximate $N(0, 1)$.

Let us repeat the above experiment, but rather than looking at the distribution of the sample means, we look at the distribution of the **standardised sample means** as they should be comparable across different sample sizes.

```python
from lets_plot import *
LetsPlot.setup_html()

import numpy as np
import pandas as pd

# Sample sizes and repetitions
n_sel_clt = [5, 10, 30, 75]  # List of sample sizes
reps = 1000  # Number of repetitions

# Population parameters
mu_x = np.mean(Xpop)  # Population mean
std_x = np.std(Xpop)  # Population standard deviation

# Container to save results
save_mean6 = np.empty((reps, len(n_sel_clt)))  # Initialize an empty matrix

# Loop through sample sizes and repetitions
for j, n_clt in enumerate(n_sel_clt):
    for i in range(reps):
        x_sample = np.random.choice(Xpop, size=n_clt, replace=True)  # Draw a sample
        xbar = np.mean(x_sample)  # Sample mean
        s_x = np.std(x_sample, ddof=1)  # Sample standard deviation
        s_xbar = s_x / np.sqrt(n_clt)  # Standard error of the mean
        save_mean6[i, j] = (xbar - mu_x) / s_xbar  # Standardized sample mean

# Convert results to a Pandas DataFrame
col_names = [f"n{n:02d}" for n in n_sel_clt]  # Create column names
results_clt2 = pd.DataFrame(save_mean6, columns=col_names)

# Reshape the DataFrame to long format
results_clt2_long = results_clt2.melt(var_name='Sample.size', value_name='sm')
```

```python
# Plotting the standardized sample means
p_clt2 = ggplot(results_clt2_long, aes(x='sm', fill='Sample.size')) + \
    geom_density(alpha=0.2) + \
    xlim(-5, 10) + \
    ggtitle("Distribution of Standardized Sample Means for Various Sample Sizes") + \
    xlab("Standardized Sample Mean") + ylab("Density")
p_clt2
```

We can see that for  larger sample sizes ($𝑛=75$), the distribution of standardized sample means closely approximates a standard normal distribution $N(0, 1)$, as the t-distribution with 74 degrees of freedom is nearly identical to a standard normal distribution. At $n=30$, the distribution is already reasonably close to normal, showing that the CLT takes a significant effect even at moderate sample sizes. However, for smaller sample sizes ($n=5$ and $n=10$), the distributions exhibit apparent asymmetries and more significant variability, reflecting the influence of the underlying population's skewness. These results highlight how the CLT ensures convergence toward normality as sample size increases while small samples remain more influenced by the original population distribution.

## The Central Limit Theorem (CLT) in Regression Context

The Central Limit Theorem (CLT) tells us that a suitably scaled average is approximately normally distributed if the sample size is sufficiently large. Interestingly, this concept applies directly to OLS regression coefficients, which are essentially scaled averages. 

Consider the simple regression model:

$$
y_i = \alpha + \beta x_i + u_i
$$

You may recall the formula for the OLS estimator of the slope coefficient $β$:

$$
\hat{β} = \frac{\text{Cov}(y_i, x_i)}{\text{Var}(x_i)}
$$

This formula demonstrates that the OLS estimator $\hat{β}$ is, in fact, "a suitably scaled average." To see why, rewrite $\hat{β}$ as:

$$
\hat{β} = \frac{1}{n} \sum_{i=1}^{n} \frac{(y_i - \overline{y})(x_i - \overline{x})}{\text{Var}(x_i)}
$$

Here, it is evident that $\hat{β}$ involves the average of $ (y_i - \overline{y})(x_i - \overline{x})$, scaled by the variance of $x_i$. Specifically, the numerator represents the covariance between $y_i$ and $x_i$, while the denominator scales this average by the variance of $x_i$.

Therefore, the CLT implies that for sufficiently large $n$, the OLS estimator $\hat{β}$ will be approximately normally distributed, as it is derived from the average of terms $(y_i - \overline{y})(x_i - \overline{x})$. This foundational result underpins much of the inferential power of regression analysis, allowing us to make statistical inferences about coefficients using standard normal approximations.

- $\hat{β}$: the OLS estimator of the slope coefficient.
- $ \text{se}(\hat{β})$ the standard error of $\hat{β}$.

Using these, we calculate the standardized t-statistic:

$$
t = \frac{\hat{\beta} - \beta}{\text{se}(\hat{\beta})}
$$

Here’s what this entails:

1. $\hat{β}$ is essentially an average of scaled terms.
2. $\text{se}(\hat{\beta})$ is the standard error, which scales $\hat{β}$ appropriately.
3. The resulting $t$-statistic is used in a t-test to evaluate the null hypothesis that the coefficient equals $β$ (which we know to be true in this case).

## Summary

The Law of Large Numbers (LLN) and the Central Limit Theorem (CLT) are fundamental principles in statistical inference, and our Python simulations effectively demonstrated their applications. The LLN ensures that as the sample size increases, the sample mean converges to the population mean, providing stable and accurate estimates. The CLT extends this by stating that, regardless of the population distribution, the distribution of suitably scaled sample means (or regression coefficients) approaches a normal distribution as the sample size grows. This was evident in our simulations of standardized sample means and regression $𝑡$-statistics, where larger sample sizes resulted in distributions closely approximating a standard normal distribution. These principles underpin hypothesis testing and regression analysis, enabling us to make robust inferences from sample data, even when the population distribution is unknown or non-normal.
