### Simple OLS Regression Model

To construct a foundational model, we perform an Ordinary Least Squares (OLS) regression to predict `lwage` using the predictors: a constant term, experience, and the logarithm of the husband's wage (`log(huswage)`). The model's equation is:

$$lwage_i = β_0 + β_1 exper_i + β_2 log(huswage_i) + u_i$$

Let's first ensure that we have the log-transformed `huswage`. This step involves ensuring that all values in `huswage` are positive. Logarithms are not defined for zero or negative values. If your dataset might contain zero or negative wages, you need to handle these cases—either by filtering them out or adjusting them accordingly!

#### Step 1: Ensure Positive Values

```python
# Filter out rows where huswage is less than or equal to zero
mydata = mydata[mydata['huswage'] > 0]
```

You can then compute the logarithm of the `huswage` column using `numpy's` log function. This will apply the natural logarithm (base e) to each value in the column:


#### Step 3: Compute the Logarithm

```python
mydata['log_huswage'] = np.log(mydata['huswage'])
```

If you expect to have zero or negative values and cannot filter them out, you might need to adjust them before applying the logarithm. One common approach is to add a small constant to all values to ensure they are positive:

#### Step 4: Handling Missing or Zero Values

```python
# Adding a small constant, for example, 1, to avoid log(0) which is undefined
mydata['log_huswage'] = np.log(mydata['huswage'] + 1)
```

This approach is often used in situations where zero represents a meaningful value (like absence of income) that you want to retain in your analysis. Adjusting by a small constant like 1 makes the logarithmic transformation feasible while minimizing the distortion to the data.

## Single coefficient hypotheses, t-tests

We use t-tests to test simple coefficient restrictions on regression coefficients. Let's model `lwage` as a function of experience and the log of the husband's wage (`log_huswage'`).

```python
# Defining the model
model = smf.ols('lwage ~ exper + log_huswage', data=mydata)
reg_ex1 = model.fit()

# Displaying the regression results
print(reg_ex1.summary())
```

As you can see, this output contains $t$-statistics and their associated p-values. These test statistics and their p-values are associated with the following hypothesis test:

$$
H_0: \beta_i = 0; \quad H_A: \beta_i \neq 0
$$

Here, $β_i$ represents the $i$-th unknown population parameter. If you want to test any other hypothesis (rather than the two-sided test for equality to 0), you will need to access the regression output to calculate the relevant test statistic.

For now, let us consider the following hypothesis:

$$
H_0: \beta_{\text{exper}} = 0.01; \quad H_A: \beta_{\text{exper}} > 0.01
$$

To perform this test, we need to calculate the test statistic:

$$
t\text{-stat} = \frac{\hat{\beta}_i - 0.01}{\text{se}(\hat{\beta}_i)}
$$


Let's now investigate the significance of individual coefficients using t-tests.

```python
print(reg_ex1.summary().tables[1])
```

Noting that the results for the `exper` variables are in the 2nd row. We can perform a one-sided t-test where the null hypothesis is that the coefficient of `exper` is equal to 0.01, and the alternative hypothesis is that it is greater than 0.01. 

```python
# Hypothesis: H0: beta_exper = 0.01; HA: beta_exper > 0.01
beta_hat = reg_ex1.params['exper']
se_beta_hat = reg_ex1.bse['exper']
t_test = (beta_hat - 0.01) / se_beta_hat

# Display the t-test result
print(f'T-test statistic: {t_test}')
```

As it turns out the value of this t-test is **1.6137**. We do not know the p-value for this particular test yet. We will turn to this shortly.

## Multiple coefficient hypotheses, F-tests

Suppose we're interested in determining the impact of adding two new variables—`age` and `educ`—to our regression model that already includes `exper` and `log_huswage`. Conducting an F-test allows us to evaluate whether these additions significantly improve the model statistically.

[//]: # (-.- .tabset)

#### Step 1: Specify Both Models
First, we need to define both the restricted model (without the additional variables) and the unrestricted model (with the additional variables).

```python
import statsmodels.formula.api as smf

# Restricted model (without 'age' and 'educ')
restricted_model = smf.ols('lwage ~ exper + log_huswage', data=mydata).fit()

# Unrestricted model (with 'age' and 'educ')
unrestricted_model = smf.ols('lwage ~ exper + log_huswage + age + educ', data=mydata).fit()
```

#### Step 2: Conduct the F-test

We then use the F-test to compare these models. The F-test examines if the explained variance in the unrestricted model is significantly greater, in proportion to the added variables, than in the restricted model.

```python
from statsmodels.stats.anova import anova_lm

# Conducting the F-test
f_test_result = anova_lm(restricted_model, unrestricted_model)
print(f_test_result)
```

As you can indicate, the output from `anova_lm` includes an F-statistic and an associated p-value. A significant p-value (typically <0.05) suggests that the additional variables (`age` and `educ`) contribute meaningfully to the model, thus justifying their inclusion. More specifically from the output of the F-test comparing the restricted and unrestricted models, you have the following key statistics:

- `df_resid`: This shows the degrees of freedom associated with the residuals from each model. For the unrestricted model (1), it is 423, indicating it has more predictors than the restricted model (0), which has 425.
  
- `ssr (sum of squared residuals)`: This indicates the total deviation of the response values from the fitted values. For the unrestricted model, it is lower (188.023144) compared to the restricted model (209.488800), suggesting a better fit as it explains more variability.

  
- `df_diff`: The difference in degrees of freedom between the models, which is 2. This corresponds to the two additional variables (age and educ) included in the unrestricted model.

  
- `ss_diff`: The difference in the sum of squared residuals between the restricted and unrestricted models, which is 21.465656. This quantifies how much more of the variance in the dependent variable is explained by including age and educ.

  
- `F`: The computed value of the F-statistic, 24.14589, measures the joint effect of age and educ on the model.

  
- `Pr(>F)`: The p-value associated with the F-test, 1.175838e-10, is extremely small, far less than any conventional significance level (e.g., 0.05, 0.01).


The very small p-value indicates that you can reject the null hypothesis that `age` and `educ` have no effect on `lwage`. In other words, adding these variables to the model provides a statistically significant improvement in model fit. Moreover, the increase in the F-statistic and the corresponding low p-value strongly suggest that including age and educ improves the model's ability to explain the variability in `lwage`.

### Calculating p-values

When calculating p-values for test statistics, several components are crucial:

- **Test Statistic Value**: You need the test statistic value you've calculated based on your sample data.
- **Distribution of the Test Statistic**: Understand the theoretical distribution of the test statistic under the null hypothesis. This will often depend on the test's assumptions about the data (e.g., normality).
- **Type of Test**: Determine whether you're conducting a two-tailed, left-tailed, or right-tailed test based on your hypothesis.

For a t-test example where the hypothesis is 
$$ H_0: \beta_{\text{exper}} = 0.01$$ 
versus 
$$ H_A: \beta_{\text{exper}} > 0.01 $$


you would perform a right-tailed test.

To calculate the p-value for a right-tailed test using a t-distribution, in Python, you can employ the `scipy.stats` library, which provides a comprehensive toolkit for statistical testing and distribution analysis. This library calculates probabilities from a t-distribution (probability under the t-distribution with 425 degrees of freedom to the left of the test-statistic t_test). The degrees of freedom is the number of observations minus the number of estimated regression coefficients (here 428-3 = 425). In our scenario, we are testing whether the coefficient of `exper` is significantly greater than 0.01. Let's calculate the p-value using Python for out t-statistic, where we are interested in the probability that the true value is greater than the calculated t-statistic under the null hypothesis.

```python
import scipy.stats as stats

# Assuming t_test is your calculated t-statistic, and df_resid is your degrees of freedom
t_test = 1.613  
df_resid = 425  

# Calculating the p-value for a right-tailed test using the survival function, which is equivalent to 1 - CDF
p_value = stats.t.sf(t_test, df=df_resid)

# Print the p-value
print(f"P-value for right-tailed test: {p_value:.3f}")
```


In Python, the approach to obtaining p-values for various statistical tests, including F-tests, similarly involves using appropriate distribution-specific functions provided by libraries such as `scipy.stats`. This suite of functions allows you to work with a broad spectrum of distributions, offering methods for both calculating critical values and determining p-values. For a list see [here](https://docs.scipy.org/doc/scipy/reference/stats.html) 

## Robust Inference in Python

In econometric analyses, tests like t-tests and F-tests typically rest on certain assumptions about the data, such as homoskedasticity, which assumes that the error terms ($u_i$) have constant variance across all observations. When these assumptions are not met—particularly with heteroskedastic data—the standard error calculations can be biased, leading to incorrect inferences.


### Heteroskedasticity Robust t-tests in Python

In econometric analysis, particularly when addressing heteroskedasticity in regression models, it's crucial to use robust standard errors to ensure accurate inference. Python provides tools for this, specifically through the `statsmodels` library. By utilisizing the `statsmodels` library, we specify a covariance type when you retrieve the model results. Let's explore how this is implemented:

[//]: # (-.- .tabset)

##### Step 1: Fit Your Regression Model
Fit your model using ordinary least squares (OLS) or another appropriate method provided by `statsmodels`.


```python
# Fit an OLS model
model = smf.ols('lwage ~ exper + log_huswage', data=mydata)
results = model.fit()
```

##### Step 2: Apply Robust Standard Errors

After fitting the model, apply robust standard errors using the `get_robustcov_results` method, specifying the type of heteroskedasticity-consistent (HC) standard errors.

```python
# Use HC1 robust standard errors
robust_results = results.get_robustcov_results(cov_type='HC1')
print(robust_results.summary())
```

Here, the `cov_type='HC1'` option is one among several available choices (e.g., `HC0`, `HC1`, `HC2`, `HC3`, `HC4`). Each type adjusts for heteroskedasticity in slightly different ways:
- `HC0`: The original White’s estimator, providing a simple correction.
- `HC1`: Adjusts the HC0 standard errors for degrees of freedom, similar to what is done in STATA and often considered a more reliable estimator under small samples.
- `HC2`: Adds further bias adjustment to the standard errors, useful in the presence of highly influential observations.
- `HC3`: More aggressively adjusts for leverage points, and is considered one of the most conservative.

As you can see, the output from `robust_results.summary()` includes the (unchanged) regression coefficients along with their new robust standard errors, t-values, and p-values.

## Heteroskedasticity and autocorrelation Robust t-tests

When error terms exhibit autocorrelation and potentially heteroskedasticity, similar concerns apply as in R, and it becomes necessary to utilize robust standard error estimators, such as the **Newey-West estimator**. In Python, using the `statsmodels` package, this can be accomplished with functions from the `statsmodels.stats.sandwich_covariance module`. 

> The equivalent function to R's `vcovHAC()` is `statsmodels.stats.sandwich_covariance.cov_hac`.

[//]: # (-.- .tabset)


#### Step 1: Model Fitting

We have already fitted a model using OLS. Here's the basic setup, for context:

```python
from statsmodels.regression.linear_model import OLS
from statsmodels.stats.sandwich_covariance import cov_hac
import statsmodels.api as sm

# Assuming 'mydata' DataFrame is already loaded and ready
X = sm.add_constant(mydata[['exper', 'log_huswage']])  # add a constant to the model
y = mydata['lwage']

model = OLS(y, X)
results = model.fit()
```

#### Step 2: Apply Newey-West Standard Errors

We will now apply Newey-West standard errors to the model you've fitted. This involves calculating the robust covariance matrix using `cov_ha`c from `statsmodels.stats.sandwich_covariance`.

```python
# Calculate the Newey-West standard errors using cov_hac
# Note: `maxlags` should be directly supplied, not in a dictionary if causing an issue
nw_cov = cov_hac(results, maxlags=1)
```

When executing this code, you might encounter the error: "TypeError: cov_hac_simple() got an unexpected keyword argument 'maxlags'". Don't let this setback deter you! Therefore, after fitting the model, use `get_robustcov_results` to adjust for autocorrelation and potential heteroskedasticity using Newey-West estimators.

```python
hac_model = model.get_robustcov_results(cov_type='HAC', maxlags=1)
```

#### Step 3: Display Results with HAC Standard Errors
Use the robust covariance matrix to display the results, including coefficients, Newey-West standard errors, t-statistics, and p-values.

```python
print(hac_model.summary())
```

## Robust F-tests


F-tests are also based on the assumption of homoskedasticity. Fortunately, addressing potential heteroskedasticity works similarly to how it is handled for robust standard errors. In Python, we can use the `statsmodels` library to specify and compute robust variance-covariance matrices and feed them into a hypothesis testing function.

We begin by specifying the same larger model as used in the standard inference section.

```python
# Fit the larger model
model = smf.ols('lwage ~ exper + np.log(huswage) + age + educ', data=mydata).fit()
```

At the core of tackling the heteroskedasticity issue is again the calculation of the correct variance-covariance matrix, which is then fed into the lht function via the `get_robustcov_results` option.

```python
# Use get_robustcov_results to calculate robust variance-covariance matrix (HC1)
robust_model = model.get_robustcov_results(cov_type='HC1')
```

The, you can perfrom the **robust F-test** for the specified null hypothesis

```python
# Check model parameters
print("Model parameters:", model.params.index)

# Define the restriction matrix
restriction_matrix = np.array([
    [0, 0, 0, 1, 0],  # Corresponds to "age = 0"
    [0, 0, 0, 0, 1]   # Corresponds to "educ = 0"
])

# Perform Wald test (equivalent to lht in R)
wald_test = robust_model.wald_test(restriction_matrix, scalar=False)

# Display the results
print(wald_test)
```


We can see that the **F-statistic (29.6)** is very large, and the **p-value (9.6e-13)** is extremely small, indicating strong evidence against the null hypothesis. The result remains unchanged, an overwhelming rejection of the null hypothesis that both age and educ are irrelevant. This means:

- Both variables (`age` and `educ`) significantly contribute to explaining the variation in `lwage` (log wage).
- The rejection of the null hypothesis suggests that omitting these variables from the model would lead to a significant loss of explanatory power.
